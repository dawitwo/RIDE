```python
import sys, os
from pathlib import Path

# Find the project root by walking up until we see a "src" folder
def find_project_root(start=None):
    p = Path(start or os.getcwd()).resolve()
    for _ in range(10):
        if (p / "src" / "config.py").exists():
            return p
        p = p.parent
    raise RuntimeError("Could not find project root (no src/config.py found)")

ROOT = find_project_root()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

print("Project root:", ROOT)
print("src exists:", (ROOT / "src").is_dir())
print("config exists:", (ROOT / "src" / "config.py").is_file())

from src.config import RAW_FILES
from src.cleaning import build_master
from src.features import add_calendar, add_weather_features, add_hist_features
```
